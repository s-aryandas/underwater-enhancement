```python
import torch
import numpy as np
from pytorch_msssim import ssim

def calculate_psnr(prediction, target, data_range=1.0):
    prediction = prediction.clamp(0, 1)
    target = target.clamp(0, 1)

    mse = torch.mean((prediction - target) ** 2, dim=(1, 2, 3))

    psnr_values = 10 * torch.log10(data_range**2 / mse.clamp(min=1e-10))

    return psnr_values


def calculate_ssim(prediction, target):
    prediction = prediction.clamp(0, 1)
    target = target.clamp(0, 1)

    ssim_values = []

    for i in range(prediction.shape[0]):
        score = ssim(
            prediction[i:i+1],
            target[i:i+1],
            data_range=1.0,
            size_average=True,
            nonnegative_ssim=True
        )

        ssim_values.append(score.item())

    return np.array(ssim_values)


print("PSNR and SSIM evaluation functions ready.")
```

```python
import pandas as pd
import torch

def evaluate_model(model, test_loader, device):
    model.eval()
    results = []

    with torch.no_grad():
        for batch_data in test_loader:
            inputs = batch_data["input"].to(device)
            targets = batch_data["target"].to(device)
            degradation = batch_data["degradation"].to(device)
            filenames = batch_data["filename"]

            predictions, branch_weights = model(inputs, degradation)

            psnr_values = calculate_psnr(predictions, targets)
            ssim_values = calculate_ssim(predictions, targets)

            for i in range(len(filenames)):
                results.append({
                    "filename": filenames[i],
                    "PSNR": psnr_values[i].item(),
                    "SSIM": ssim_values[i]
                })

    results_df = pd.DataFrame(results)

    summary = {
        "Mean_PSNR": results_df["PSNR"].mean(),
        "Mean_SSIM": results_df["SSIM"].mean(),
        "Images_Evaluated": len(results_df)
    }

    return results_df, summary


print("Test evaluation function ready.")
```

```python
import numpy as np
import pandas as pd
import torch
from skimage.color import rgb2hsv, hsv2rgb
from skimage.exposure import equalize_adapthist


def gray_world_tensor(image):
    """
    Gray World white balance for a single image tensor (C,H,W).
    """
    image = image.clamp(0, 1)

    channel_means = image.mean(dim=(1, 2), keepdim=True)
    gray_mean = channel_means.mean()

    adjusted = image * (gray_mean / channel_means.clamp(min=1e-6))

    return adjusted.clamp(0, 1)


def clahe_tensor(image, clip_limit=0.01):
    """
    Apply CLAHE to the HSV Value channel of a single image tensor.
    Input: (C,H,W), RGB, range [0,1].
    """
    image_np = image.detach().cpu().permute(1, 2, 0).numpy()
    image_np = np.clip(image_np, 0, 1)

    hsv = rgb2hsv(image_np)

    hsv[:, :, 2] = equalize_adapthist(
        hsv[:, :, 2],
        clip_limit=clip_limit
    )

    enhanced = hsv2rgb(hsv)

    enhanced = np.clip(enhanced, 0, 1)

    return torch.from_numpy(enhanced).permute(2, 0, 1).float()


def evaluate_classical_baselines(test_loader):
    methods = {
        "GrayWorld": gray_world_tensor,
        "CLAHE": clahe_tensor
    }

    results = []

    for batch_data in test_loader:
        inputs = batch_data["input"]
        targets = batch_data["target"]
        filenames = batch_data["filename"]

        for method_name, method in methods.items():
            predictions = torch.stack([
                method(image) for image in inputs
            ])

            psnr_values = calculate_psnr(predictions, targets)
            ssim_values = calculate_ssim(predictions, targets)

            for i, filename in enumerate(filenames):
                results.append({
                    "filename": filename,
                    "Method": method_name,
                    "PSNR": psnr_values[i].item(),
                    "SSIM": ssim_values[i]
                })

    baseline_df = pd.DataFrame(results)

    summary_df = (
        baseline_df.groupby("Method")[["PSNR", "SSIM"]]
        .mean()
        .reset_index()
    )

    return baseline_df, summary_df


print("Classical baseline evaluation functions ready.")
```

```python
import cv2
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
from pathlib import Path

IMAGE_SIZE = 256
BATCH_SIZE = 8

# ---------- Feature extraction ----------
def extract_degradation_features(image):
    """
    image: RGB image as numpy array, range 0-255
    Returns:
        color imbalance, global contrast, log Laplacian variance
    """
    image_float = image.astype(np.float32) / 255.0

    channel_means = image_float.mean(axis=(0, 1))
    color_imbalance = np.std(channel_means)

    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    global_contrast = float(gray.std())

    laplacian_variance = float(cv2.Laplacian(gray, cv2.CV_64F).var())
    log_laplacian = float(np.log1p(laplacian_variance))

    return np.array([
        color_imbalance,
        global_contrast,
        log_laplacian
    ], dtype=np.float32)


# ---------- Dataset ----------
class AquaLumeDataset(Dataset):
    def __init__(self, filenames, manifest, raw_dir, reference_dir, scaler):
        self.filenames = list(filenames)
        self.manifest = manifest
        self.raw_dir = Path(raw_dir)
        self.reference_dir = Path(reference_dir)
        self.scaler = scaler

    def __len__(self):
        return len(self.filenames)

    def __getitem__(self, idx):
        filename = self.filenames[idx]

        raw_path = self.raw_dir / filename
        ref_path = self.reference_dir / filename

        raw_image = cv2.imread(str(raw_path))
        ref_image = cv2.imread(str(ref_path))

        if raw_image is None or ref_image is None:
            raise FileNotFoundError(f"Could not load image pair: {filename}")

        raw_image = cv2.cvtColor(raw_image, cv2.COLOR_BGR2RGB)
        ref_image = cv2.cvtColor(ref_image, cv2.COLOR_BGR2RGB)

        raw_image = cv2.resize(raw_image, (IMAGE_SIZE, IMAGE_SIZE))
        ref_image = cv2.resize(ref_image, (IMAGE_SIZE, IMAGE_SIZE))

        features = extract_degradation_features(raw_image)
        features = self.scaler.transform(features.reshape(1, -1))[0]

        input_tensor = torch.from_numpy(
            raw_image.astype(np.float32) / 255.0
        ).permute(2, 0, 1)

        target_tensor = torch.from_numpy(
            ref_image.astype(np.float32) / 255.0
        ).permute(2, 0, 1)

        degradation_tensor = torch.tensor(features, dtype=torch.float32)

        return {
            "input": input_tensor,
            "target": target_tensor,
            "degradation": degradation_tensor,
            "filename": filename
        }


# ---------- Corrected split ----------
train_filenames = usable_manifest[
    usable_manifest["split"] == "train"
]["filename"].tolist()

val_filenames = usable_manifest[
    usable_manifest["split"] == "validation"
]["filename"].tolist()

test_filenames = usable_manifest[
    usable_manifest["split"] == "test"
]["filename"].tolist()


# ---------- Fit scaler ONLY on training images ----------
train_raw_features = []

for filename in train_filenames:
    image = cv2.imread(str(RAW_DIR / filename))

    if image is None:
        raise FileNotFoundError(filename)

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (IMAGE_SIZE, IMAGE_SIZE))

    train_raw_features.append(extract_degradation_features(image))

train_raw_features = np.asarray(train_raw_features)

feature_scaler = StandardScaler()
feature_scaler.fit(train_raw_features)

print("Scaler fitted using training images only.")
print("Scaler mean:", feature_scaler.mean_)
print("Scaler std:", feature_scaler.scale_)


# ---------- Create datasets ----------
train_dataset = AquaLumeDataset(
    train_filenames, usable_manifest, RAW_DIR, REF_DIR, feature_scaler
)

val_dataset = AquaLumeDataset(
    val_filenames, usable_manifest, RAW_DIR, REF_DIR, feature_scaler
)

test_dataset = AquaLumeDataset(
    test_filenames, usable_manifest, RAW_DIR, REF_DIR, feature_scaler
)


# ---------- Create DataLoaders ----------
train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE,
    shuffle=True, num_workers=2, pin_memory=True
)

val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE,
    shuffle=False, num_workers=2, pin_memory=True
)

test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE,
    shuffle=False, num_workers=2, pin_memory=True
)


print("\nDataLoaders created successfully!")
print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn

class AquaLumeCNN(nn.Module):
    def __init__(self, input_channels=3, features=64):
        super().__init__()

        self.features = nn.Sequential(
            # Stage 1
            nn.Conv2d(input_channels, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),

            # Stage 2
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),

            # Stage 3
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.features(x)


# CNN branch smoke test
cnn = AquaLumeCNN().to(DEVICE)

dummy_input = torch.randn(2, 3, IMAGE_SIZE, IMAGE_SIZE).to(DEVICE)
cnn_output = cnn(dummy_input)

print("CNN input shape:", dummy_input.shape)
print("CNN output shape:", cnn_output.shape)
print("CNN parameters:", sum(p.numel() for p in cnn.parameters()))
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AquaLumeViT(nn.Module):
    def __init__(
        self,
        image_size=256,
        patch_size=16,
        input_channels=3,
        embed_dim=64,
        num_layers=2,
        num_heads=4
    ):
        super().__init__()

        self.patch_embed = nn.Conv2d(
            input_channels,
            embed_dim,
            kernel_size=patch_size,
            stride=patch_size
        )

        num_patches = (image_size // patch_size) ** 2

        self.pos_embedding = nn.Parameter(
            torch.randn(1, num_patches, embed_dim) * 0.02
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim,
            nhead=num_heads,
            dim_feedforward=128,
            dropout=0.1,
            batch_first=True,
            activation="gelu"
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

    def forward(self, x):
        # Image -> patch embeddings
        x = self.patch_embed(x)

        # (B, 64, 16, 16) -> (B, 256, 64)
        x = x.flatten(2).transpose(1, 2)

        # Add positional information
        x = x + self.pos_embedding

        # Global contextual processing
        x = self.transformer(x)

        # (B, 256, 64) -> (B, 64, 16, 16)
        x = x.transpose(1, 2).reshape(
            x.shape[0], 64, 16, 16
        )

        # Match the CNN feature spatial resolution
        x = F.interpolate(
            x,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        return x


# ViT branch smoke test
vit = AquaLumeViT().to(DEVICE)

dummy_input = torch.randn(2, 3, 256, 256).to(DEVICE)

with torch.no_grad():
    vit_output = vit(dummy_input)

print("ViT input shape:", dummy_input.shape)
print("ViT output shape:", vit_output.shape)
print("ViT parameters:", sum(p.numel() for p in vit.parameters()))
print("GPU memory allocated:", round(torch.cuda.memory_allocated()/1024**2, 2), "MB")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AquaLumePhysics(nn.Module):
    def __init__(self, input_channels=3, degradation_dim=3, features=64):
        super().__init__()

        # Image feature extraction
        self.image_encoder = nn.Sequential(
            nn.Conv2d(input_channels, 32, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.ReLU(inplace=True)
        )

        # Degradation descriptor encoding
        self.degradation_encoder = nn.Sequential(
            nn.Linear(degradation_dim, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 32),
            nn.ReLU(inplace=True)
        )

        # Image + degradation feature fusion
        self.fusion = nn.Sequential(
            nn.Conv2d(64, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, kernel_size=3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, image, degradation):
        # Resize image for efficient feature extraction
        image = F.interpolate(
            image,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        image_features = self.image_encoder(image)

        degradation_features = self.degradation_encoder(degradation)

        # Spatial broadcasting
        degradation_features = degradation_features.unsqueeze(-1).unsqueeze(-1)
        degradation_features = degradation_features.expand(
            -1, -1, 64, 64
        )

        combined = torch.cat(
            [image_features, degradation_features],
            dim=1
        )

        return self.fusion(combined)


# Smoke test
physics_branch = AquaLumePhysics().to(DEVICE)

dummy_image = torch.randn(2, 3, 256, 256).to(DEVICE)
dummy_degradation = torch.randn(2, 3).to(DEVICE)

with torch.no_grad():
    physics_output = physics_branch(dummy_image, dummy_degradation)

print("Physics input image:", dummy_image.shape)
print("Degradation input:", dummy_degradation.shape)
print("Physics output:", physics_output.shape)
print("Physics parameters:", sum(p.numel() for p in physics_branch.parameters()))
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AdaptiveFusion(nn.Module):
    def __init__(self, degradation_dim=3, features=64):
        super().__init__()

        self.weight_generator = nn.Sequential(
            nn.Linear(degradation_dim, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 3)
        )

        self.fusion_conv = nn.Sequential(
            nn.Conv2d(features * 3, features, 1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, cnn_features, vit_features, physics_features, degradation):

        weights = F.softmax(
            self.weight_generator(degradation),
            dim=1
        )

        w_cnn = weights[:, 0].view(-1, 1, 1, 1)
        w_vit = weights[:, 1].view(-1, 1, 1, 1)
        w_physics = weights[:, 2].view(-1, 1, 1, 1)

        cnn_weighted = cnn_features * w_cnn
        vit_weighted = vit_features * w_vit
        physics_weighted = physics_features * w_physics

        combined = torch.cat(
            [cnn_weighted, vit_weighted, physics_weighted],
            dim=1
        )

        fused = self.fusion_conv(combined)

        return fused, weights


class AquaLume(nn.Module):
    def __init__(self):
        super().__init__()

        self.cnn = AquaLumeCNN()
        self.vit = AquaLumeViT()
        self.physics = AquaLumePhysics()
        self.adaptive_fusion = AdaptiveFusion()

        self.decoder = nn.Sequential(
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Upsample(scale_factor=2, mode="bilinear", align_corners=False),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Upsample(scale_factor=2, mode="bilinear", align_corners=False),
            nn.Conv2d(64, 32, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 3, 3, padding=1),
            nn.Sigmoid()
        )

    def forward(self, image, degradation):

        # CNN local feature branch
        cnn_features = self.cnn(image)

        # Adapt CNN features to common spatial resolution
        cnn_features = F.interpolate(
            cnn_features,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        # ViT global feature branch
        vit_features = self.vit(image)

        # Physics-inspired degradation-aware branch
        physics_features = self.physics(image, degradation)

        # Adaptive three-branch fusion
        fused_features, weights = self.adaptive_fusion(
            cnn_features,
            vit_features,
            physics_features,
            degradation
        )

        # Image reconstruction
        enhanced_image = self.decoder(fused_features)

        return enhanced_image, weights


# Instantiate model
model = AquaLume().to(DEVICE)

# Smoke test
dummy_image = torch.randn(2, 3, 256, 256).to(DEVICE)
dummy_degradation = torch.randn(2, 3).to(DEVICE)

with torch.no_grad():
    output, branch_weights = model(dummy_image, dummy_degradation)

print("Input:", dummy_image.shape)
print("Enhanced output:", output.shape)
print("Branch weights:", branch_weights.shape)
print("Sample branch weights:", branch_weights[0].cpu().numpy())
print("Weight sum:", branch_weights[0].sum().item())

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print("GPU memory allocated:", round(torch.cuda.memory_allocated()/1024**2, 2), "MB")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from pytorch_msssim import ssim

class AquaLumeLoss(nn.Module):
    def __init__(self):
        super().__init__()
        self.l1 = nn.L1Loss()

    def gradient_loss(self, prediction, target):
        pred_dx = prediction[:, :, :, 1:] - prediction[:, :, :, :-1]
        pred_dy = prediction[:, :, 1:, :] - prediction[:, :, :-1, :]

        target_dx = target[:, :, :, 1:] - target[:, :, :, :-1]
        target_dy = target[:, :, 1:, :] - target[:, :, :-1, :]

        loss_x = F.l1_loss(pred_dx, target_dx)
        loss_y = F.l1_loss(pred_dy, target_dy)

        return loss_x + loss_y

    def forward(self, prediction, target):
        l1_loss = self.l1(prediction, target)

        ssim_value = ssim(
            prediction,
            target,
            data_range=1.0,
            size_average=True,
            nonnegative_ssim=True
        )

        ssim_loss = 1.0 - ssim_value

        grad_loss = self.gradient_loss(prediction, target)

        total_loss = l1_loss + 0.5 * ssim_loss + 0.1 * grad_loss

        return total_loss, {
            "L1": l1_loss.detach(),
            "SSIM": ssim_value.detach(),
            "Gradient": grad_loss.detach()
        }


criterion = AquaLumeLoss().to(DEVICE)

# Test loss and gradient flow
dummy_prediction = torch.rand(2, 3, 256, 256).to(DEVICE)
dummy_target = torch.rand(2, 3, 256, 256).to(DEVICE)

loss_value, loss_components = criterion(dummy_prediction, dummy_target)

print("Total loss:", loss_value.item())
print("Loss components:", {
    key: value.item() for key, value in loss_components.items()
})
```

```python
import time
import copy
import json
import torch

start_time = time.time()

best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0

for epoch in range(NUM_EPOCHS):

    epoch_start = time.time()

    train_loss = run_training_epoch(
        model, train_loader, criterion, optimizer, DEVICE
    )

    val_loss = run_validation_epoch(
        model, val_loader, criterion, DEVICE
    )

    scheduler.step(val_loss)

    current_lr = optimizer.param_groups[0]["lr"]

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["lr"].append(current_lr)

    improved = val_loss < best_val_loss

    if improved:
        best_val_loss = val_loss
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_loss": val_loss,
            "train_loss": train_loss,
        }, CHECKPOINT_DIR / "aqualume_best_model.pth")

    else:
        epochs_without_improvement += 1

    elapsed_epoch = time.time() - epoch_start

    print(
        f"Epoch [{epoch+1:02d}/{NUM_EPOCHS}] | "
        f"Train: {train_loss:.5f} | "
        f"Val: {val_loss:.5f} | "
        f"LR: {current_lr:.2e} | "
        f"Time: {elapsed_epoch:.1f}s | "
        f"{'BEST' if improved else ''}"
    )

    if epochs_without_improvement >= PATIENCE:
        print(f"\nEarly stopping at epoch {epoch+1}.")
        break

# Save training history
with open(LOG_DIR / "training_history.json", "w") as f:
    json.dump(history, f, indent=4)

# Restore best model
checkpoint = torch.load(
    CHECKPOINT_DIR / "aqualume_best_model.pth",
    map_location=DEVICE,
    weights_only=False
)

model.load_state_dict(checkpoint["model_state_dict"])

print("\n" + "="*50)
print("TRAINING COMPLETED")
print("Best epoch:", best_epoch)
print("Best validation loss:", best_val_loss)
print("Total training time:", round((time.time()-start_time)/60, 2), "minutes")
print("Best model restored.")
print("="*50)
```

```python
import torch
import torch.nn as nn
from torch.optim import Adam
from torch.optim.lr_scheduler import ReduceLROnPlateau
import time
import copy

optimizer = Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

def run_training_epoch(model, loader, criterion, optimizer, device):
    model.train()
    total_loss = 0.0
    batches = 0

    for batch in loader:
        inputs = batch["input"].to(device)
        targets = batch["target"].to(device)
        degradation = batch["degradation"].to(device)

        optimizer.zero_grad(set_to_none=True)

        predictions, branch_weights = model(inputs, degradation)
        loss, components = criterion(predictions, targets)

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        total_loss += loss.item()
        batches += 1

    return total_loss / max(batches, 1)


@torch.no_grad()
def run_validation_epoch(model, loader, criterion, device):
    model.eval()
    total_loss = 0.0
    batches = 0

    for batch in loader:
        inputs = batch["input"].to(device)
        targets = batch["target"].to(device)
        degradation = batch["degradation"].to(device)

        predictions, branch_weights = model(inputs, degradation)
        loss, components = criterion(predictions, targets)

        total_loss += loss.item()
        batches += 1

    return total_loss / max(batches, 1)


history = {
    "train_loss": [],
    "val_loss": [],
    "lr": []
}

best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0

print("Training infrastructure ready.")
print("Optimizer: Adam")
print("Scheduler: ReduceLROnPlateau")
print("Early stopping patience:", PATIENCE)
```

```python
import numpy as np
import pandas as pd
import cv2
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
from pathlib import Path

RAW_DIR = Path("/content/datasets/UIEB/UIEB/Raw")
REF_DIR = Path("/content/datasets/UIEB/UIEB/References")

IMAGE_SIZE = 256
BATCH_SIZE = 8

# --------------------------------------------------
# 1. Calculate degradation descriptors
# --------------------------------------------------
def calculate_degradation_features(image_path):

    image = cv2.imread(str(image_path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # RGB channel mean imbalance
    channel_means = image.mean(axis=(0, 1)) / 255.0
    color_imbalance = np.std(channel_means)

    # Global grayscale contrast
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    contrast = np.std(gray)

    # Log-Laplacian variance
    laplacian_variance = cv2.Laplacian(
        gray, cv2.CV_64F
    ).var()

    sharpness = np.log1p(laplacian_variance)

    return np.array([
        color_imbalance,
        contrast,
        sharpness
    ], dtype=np.float32)


# --------------------------------------------------
# 2. Calculate features for the corrected manifest
# --------------------------------------------------
usable_manifest = usable_manifest.reset_index(drop=True)

feature_records = []

for filename in usable_manifest["filename"]:
    path = RAW_DIR / filename

    if not path.exists():
        raise FileNotFoundError(f"Missing raw image: {path}")

    feature_records.append(calculate_degradation_features(path))

raw_features = np.array(feature_records, dtype=np.float32)

print("Raw degradation features:", raw_features.shape)


# --------------------------------------------------
# 3. Fit scaler on training data only
# --------------------------------------------------
train_mask = usable_manifest["split"].values == "train"

feature_scaler = StandardScaler()
feature_scaler.fit(raw_features[train_mask])

scaled_features = feature_scaler.transform(raw_features).astype(np.float32)

print("Scaler mean:", feature_scaler.mean_)
print("Scaler std:", feature_scaler.scale_)


# --------------------------------------------------
# 4. Dataset
# --------------------------------------------------
class AquaLumeDataset(Dataset):

    def __init__(self, manifest, scaled_features, raw_dir, ref_dir):
        self.manifest = manifest.reset_index(drop=True)
        self.features = scaled_features
        self.raw_dir = raw_dir
        self.ref_dir = ref_dir

    def __len__(self):
        return len(self.manifest)

    def __getitem__(self, idx):

        filename = self.manifest.iloc[idx]["filename"]

        raw_path = self.raw_dir / filename
        ref_path = self.ref_dir / filename

        if not ref_path.exists():
            raise FileNotFoundError(f"Missing reference: {ref_path}")

        input_image = Image.open(raw_path).convert("RGB")
        target_image = Image.open(ref_path).convert("RGB")

        input_image = input_image.resize(
            (IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.BILINEAR
        )
        target_image = target_image.resize(
            (IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.BILINEAR
        )

        input_array = np.asarray(input_image, dtype=np.float32) / 255.0
        target_array = np.asarray(target_image, dtype=np.float32) / 255.0

        input_tensor = torch.from_numpy(
            input_array.transpose(2, 0, 1).copy()
        )

        target_tensor = torch.from_numpy(
            target_array.transpose(2, 0, 1).copy()
        )

        degradation_tensor = torch.from_numpy(
            self.features[idx].copy()
        )

        return {
            "input": input_tensor,
            "target": target_tensor,
            "degradation": degradation_tensor,
            "filename": filename
        }


# --------------------------------------------------
# 5. Create split datasets and loaders
# --------------------------------------------------
train_indices = usable_manifest.index[
    usable_manifest["split"] == "train"
].tolist()

val_indices = usable_manifest.index[
    usable_manifest["split"] == "validation"
].tolist()

test_indices = usable_manifest.index[
    usable_manifest["split"] == "test"
].tolist()


train_manifest = usable_manifest.iloc[train_indices].reset_index(drop=True)
val_manifest = usable_manifest.iloc[val_indices].reset_index(drop=True)
test_manifest = usable_manifest.iloc[test_indices].reset_index(drop=True)


train_features = scaled_features[train_indices]
val_features = scaled_features[val_indices]
test_features = scaled_features[test_indices]


train_dataset = AquaLumeDataset(
    train_manifest, train_features, RAW_DIR, REF_DIR
)

val_dataset = AquaLumeDataset(
    val_manifest, val_features, RAW_DIR, REF_DIR
)

test_dataset = AquaLumeDataset(
    test_manifest, test_features, RAW_DIR, REF_DIR
)


train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=2, pin_memory=True
)

val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=2, pin_memory=True
)

test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=2, pin_memory=True
)


# --------------------------------------------------
# 6. Verification
# --------------------------------------------------
print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))

batch = next(iter(train_loader))

print("\nBatch verification:")
print("Input:", batch["input"].shape)
print("Target:", batch["target"].shape)
print("Degradation:", batch["degradation"].shape)
print("Filename example:", batch["filename"][0])

print("\nData pipeline successfully rebuilt.")
```

```python
import cv2
import numpy as np
import pandas as pd
import torch

from skimage.metrics import (
    peak_signal_noise_ratio,
    structural_similarity
)
from tqdm.auto import tqdm

# ------------------------------
# BASELINE METHODS
# ------------------------------

def gray_world(image):
    """
    image: RGB float image in [0,1]
    """
    means = image.mean(axis=(0, 1))
    overall_mean = means.mean()

    scales = overall_mean / (means + 1e-8)
    corrected = image * scales

    return np.clip(corrected, 0, 1)


def apply_clahe(image):
    """
    CLAHE on LAB luminance.
    RGB float image in [0,1].
    """
    img_uint8 = np.clip(image * 255, 0, 255).astype(np.uint8)

    lab = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2LAB)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    lab[:, :, 0] = clahe.apply(lab[:, :, 0])

    enhanced = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)

    return enhanced.astype(np.float32) / 255.0


# ------------------------------
# METRIC CALCULATION
# ------------------------------

def calculate_metrics(pred, target):

    pred = np.clip(pred, 0, 1)
    target = np.clip(target, 0, 1)

    mae = np.mean(np.abs(pred - target))
    rmse = np.sqrt(np.mean((pred - target) ** 2))

    psnr = peak_signal_noise_ratio(
        target, pred, data_range=1.0
    )

    ssim = structural_similarity(
        target,
        pred,
        data_range=1.0,
        channel_axis=-1
    )

    return psnr, ssim, mae, rmse


# ------------------------------
# EVALUATION
# ------------------------------

model.eval()

baseline_results = []

with torch.no_grad():

    for batch in tqdm(test_loader, desc="Baseline comparison"):

        inputs = batch["input"].to(device)
        targets = batch["target"].to(device)
        degradation = batch["degradation"].to(device)

        predictions, branch_weights = model(
            inputs, degradation
        )

        inputs_np = inputs.cpu().numpy()
        targets_np = targets.cpu().numpy()
        predictions_np = predictions.clamp(0, 1).cpu().numpy()

        filenames = batch["filename"]

        for i, filename in enumerate(filenames):

            original = np.transpose(
                inputs_np[i], (1, 2, 0)
            )

            target = np.transpose(
                targets_np[i], (1, 2, 0)
            )

            aqualume = np.transpose(
                predictions_np[i], (1, 2, 0)
            )

            gray_world_img = gray_world(original)
            clahe_img = apply_clahe(original)

            methods = {
                "Original": original,
                "GrayWorld": gray_world_img,
                "CLAHE": clahe_img,
                "AquaLume": aqualume
            }

            for method_name, image in methods.items():

                psnr, ssim, mae, rmse = calculate_metrics(
                    image, target
                )

                baseline_results.append({
                    "filename": filename,
                    "Method": method_name,
                    "PSNR": psnr,
                    "SSIM": ssim,
                    "MAE": mae,
                    "RMSE": rmse
                })


# ------------------------------
# RESULTS TABLE
# ------------------------------

baseline_df = pd.DataFrame(baseline_results)

summary = (
    baseline_df
    .groupby("Method")[["PSNR", "SSIM", "MAE", "RMSE"]]
    .mean()
    .sort_values("PSNR", ascending=False)
)

print("\nTEST SET BASELINE COMPARISON")
print("=" * 70)
print(summary.round(4))

# Save detailed and summary results
base_dir = "/content/drive/MyDrive/AquaLume/trained_model"

baseline_df.to_csv(
    f"{base_dir}/baseline_per_image_results.csv",
    index=False
)

summary.to_csv(
    f"{base_dir}/baseline_summary.csv"
)

print("\nDetailed results saved.")
print("Summary saved.")
```

```python
import torch
import numpy as np
from PIL import Image

# =========================
# AQUAlume INFERENCE
# =========================

model.eval()
model.to(device)

def enhance_underwater(image, return_weights=False):
    """
    Enhances an underwater image using trained AquaLume.

    Parameters
    ----------
    image : PIL.Image or numpy.ndarray
        RGB image.
    return_weights : bool
        Whether to return adaptive branch weights.

    Returns
    -------
    enhanced : numpy.ndarray
        Enhanced RGB image in [0,1].
    weights : numpy.ndarray, optional
        CNN, ViT and physics branch weights.
    """

    # Convert input to RGB numpy
    if isinstance(image, Image.Image):
        image = image.convert("RGB")
        image = np.asarray(image)

    image = np.asarray(image)

    if image.dtype == np.uint8:
        image = image.astype(np.float32) / 255.0
    else:
        image = image.astype(np.float32)

        if image.max() > 1.0:
            image = image / 255.0

    image = np.clip(image, 0, 1)

    # Resize to model input size
    image_resized = Image.fromarray(
        (image * 255).astype(np.uint8)
    ).resize((256, 256), Image.Resampling.BILINEAR)

    image_resized = np.asarray(
        image_resized
    ).astype(np.float32) / 255.0

    # Tensor conversion
    input_tensor = torch.from_numpy(
        image_resized.transpose(2, 0, 1)
    ).unsqueeze(0).float().to(device)

    # Compute degradation features using existing pipeline function
    degradation_features = compute_degradation_features(
        image_resized
    )

    degradation_tensor = torch.tensor(
        degradation_features,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    # Inference
    with torch.no_grad():

        prediction, branch_weights = model(
            input_tensor,
            degradation_tensor
        )

    prediction = prediction.squeeze(0).cpu().numpy()
    prediction = np.transpose(prediction, (1, 2, 0))

    prediction = np.clip(prediction, 0, 1)

    weights = branch_weights.squeeze(0).cpu().numpy()

    if return_weights:
        return prediction, weights

    return prediction


print("AquaLume standalone inference function created.")
```

```python
import numpy as np
import cv2

def compute_degradation_features(image):
    """
    Extract the three degradation features used by the
    recovered AquaLume preprocessing pipeline and apply
    the training-fitted scaler.

    image: RGB image, float in [0,1], resized to 256x256.
    """

    image = np.asarray(image, dtype=np.float32)
    image = np.clip(image, 0, 1)

    # 1. RGB channel-mean imbalance
    channel_means = image.mean(axis=(0, 1))
    color_imbalance = np.std(channel_means)

    # Convert RGB to grayscale in 8-bit range
    image_uint8 = (image * 255).astype(np.uint8)
    gray = cv2.cvtColor(image_uint8, cv2.COLOR_RGB2GRAY)

    # 2. Global grayscale contrast
    contrast = np.std(gray)

    # 3. Laplacian sharpness
    laplacian_variance = cv2.Laplacian(
        gray, cv2.CV_64F
    ).var()

    sharpness = np.log1p(laplacian_variance)

    raw_features = np.array(
        [color_imbalance, contrast, sharpness],
        dtype=np.float32
    ).reshape(1, -1)

    # Apply the scaler fitted on training data
    scaled_features = feature_scaler.transform(raw_features)

    return scaled_features.squeeze(0).astype(np.float32)


print("Degradation feature extraction restored.")
print("Feature scaler:", type(feature_scaler).__name__)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from pytorch_msssim import ssim

class AquaLumeV2Loss(nn.Module):

    def __init__(self):
        super().__init__()

        self.l1 = nn.L1Loss()

    def gradient_loss(self, pred, target):

        pred_dx = pred[:, :, :, 1:] - pred[:, :, :, :-1]
        target_dx = target[:, :, :, 1:] - target[:, :, :, :-1]

        pred_dy = pred[:, :, 1:, :] - pred[:, :, :-1, :]
        target_dy = target[:, :, 1:, :] - target[:, :, :-1, :]

        loss_x = F.l1_loss(pred_dx, target_dx)
        loss_y = F.l1_loss(pred_dy, target_dy)

        return loss_x + loss_y

    def edge_loss(self, pred, target):

        # Fixed Laplacian filter
        kernel = torch.tensor(
            [[0., 1., 0.],
             [1., -4., 1.],
             [0., 1., 0.]],
            device=pred.device,
            dtype=pred.dtype
        ).view(1, 1, 3, 3)

        # Apply to each RGB channel independently
        kernel = kernel.repeat(3, 1, 1, 1)

        pred_edges = F.conv2d(
            pred, kernel, padding=1, groups=3
        )

        target_edges = F.conv2d(
            target, kernel, padding=1, groups=3
        )

        return F.l1_loss(pred_edges, target_edges)

    def forward(self, pred, target):

        l1_loss = self.l1(pred, target)

        ssim_loss = 1 - ssim(
            pred,
            target,
            data_range=1.0,
            size_average=True
        )

        grad_loss = self.gradient_loss(pred, target)

        edge_loss = self.edge_loss(pred, target)

        total_loss = (
            1.0 * l1_loss
            + 0.75 * ssim_loss
            + 0.30 * grad_loss
            + 0.15 * edge_loss
        )

        components = {
            "L1": l1_loss.item(),
            "SSIM": ssim_loss.item(),
            "Gradient": grad_loss.item(),
            "Edge": edge_loss.item()
        }

        return total_loss, components


criterion_v2 = AquaLumeV2Loss().to(device)

print("AquaLume v2 detail-preserving loss initialized.")
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F

class CNNBranch(nn.Module):
    def __init__(self, channels=64):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.features(x)


class ViTBranch(nn.Module):
    def __init__(self, embed_dim=64, patch_size=16):
        super().__init__()

        self.patch_embed = nn.Conv2d(
            3, embed_dim,
            kernel_size=patch_size,
            stride=patch_size
        )

        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=embed_dim,
                nhead=4,
                dim_feedforward=128,
                batch_first=True
            ),
            num_layers=2
        )

    def forward(self, x):
        patches = self.patch_embed(x)

        B, C, H, W = patches.shape
        tokens = patches.flatten(2).transpose(1, 2)

        tokens = self.transformer(tokens)

        features = tokens.transpose(1, 2).reshape(B, C, H, W)

        return features


class PhysicsBranch(nn.Module):
    def __init__(self, channels=64):
        super().__init__()

        self.image_encoder = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        self.degradation_encoder = nn.Sequential(
            nn.Linear(3, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 32),
            nn.ReLU(inplace=True)
        )

        self.fusion = nn.Sequential(
            nn.Conv2d(64, channels, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x, degradation):
        image_features = self.image_encoder(x)

        degradation_features = self.degradation_encoder(degradation)

        degradation_features = degradation_features.unsqueeze(-1).unsqueeze(-1)

        degradation_features = degradation_features.expand(
            -1, -1, image_features.shape[2], image_features.shape[3]
        )

        fused = torch.cat([image_features, degradation_features], dim=1)

        return self.fusion(fused)


class AquaLume(nn.Module):
    def __init__(self):
        super().__init__()

        self.cnn = CNNBranch(64)
        self.vit = ViTBranch(64, 16)
        self.physics = PhysicsBranch(64)

        self.weight_generator = nn.Sequential(
            nn.Linear(3, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 3)
        )

        self.decoder = nn.Sequential(
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Upsample(scale_factor=2, mode="bilinear", align_corners=False),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Upsample(scale_factor=2, mode="bilinear", align_corners=False),
            nn.Conv2d(64, 3, 3, padding=1),
            nn.Sigmoid()
        )

    def forward(self, x, degradation):
        cnn_features = self.cnn(x)
        vit_features = self.vit(x)
        physics_features = self.physics(x, degradation)

        cnn_features = F.interpolate(
            cnn_features, size=(64, 64), mode="bilinear", align_corners=False
        )

        weights = torch.softmax(
            self.weight_generator(degradation), dim=1
        )

        fused = (
            weights[:, 0].view(-1, 1, 1, 1) * cnn_features +
            weights[:, 1].view(-1, 1, 1, 1) * vit_features +
            weights[:, 2].view(-1, 1, 1, 1) * physics_features
        )

        output = self.decoder(fused)

        return output, weights
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# =========================================================
# 1. CNN BRANCH
# =========================================================
class AquaLumeCNN(nn.Module):
    def __init__(self, input_channels=3, features=64):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(input_channels, features, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.features(x)


# =========================================================
# 2. ViT BRANCH
# =========================================================
class AquaLumeViT(nn.Module):
    def __init__(
        self,
        image_size=256,
        patch_size=16,
        input_channels=3,
        embed_dim=64,
        num_layers=2,
        num_heads=4
    ):
        super().__init__()

        self.patch_embed = nn.Conv2d(
            input_channels,
            embed_dim,
            kernel_size=patch_size,
            stride=patch_size
        )

        num_patches = (image_size // patch_size) ** 2

        self.pos_embedding = nn.Parameter(
            torch.randn(1, num_patches, embed_dim) * 0.02
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim,
            nhead=num_heads,
            dim_feedforward=128,
            dropout=0.1,
            batch_first=True,
            activation="gelu"
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

    def forward(self, x):
        x = self.patch_embed(x)

        x = x.flatten(2).transpose(1, 2)

        x = x + self.pos_embedding

        x = self.transformer(x)

        x = x.transpose(1, 2).reshape(
            x.shape[0], 64, 16, 16
        )

        x = F.interpolate(
            x,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        return x


# =========================================================
# 3. PHYSICS / DEGRADATION-AWARE BRANCH
# =========================================================
class AquaLumePhysics(nn.Module):
    def __init__(self, input_channels=3, degradation_dim=3, features=64):
        super().__init__()

        self.image_encoder = nn.Sequential(
            nn.Conv2d(input_channels, 32, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        self.degradation_encoder = nn.Sequential(
            nn.Linear(degradation_dim, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 32),
            nn.ReLU(inplace=True)
        )

        self.fusion = nn.Sequential(
            nn.Conv2d(64, features, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, image, degradation):
        image = F.interpolate(
            image,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        image_features = self.image_encoder(image)

        degradation_features = self.degradation_encoder(degradation)

        degradation_features = (
            degradation_features.unsqueeze(-1).unsqueeze(-1)
        )

        degradation_features = degradation_features.expand(
            -1, -1, 64, 64
        )

        combined = torch.cat(
            [image_features, degradation_features],
            dim=1
        )

        return self.fusion(combined)


# =========================================================
# 4. ADAPTIVE FUSION
# =========================================================
class AdaptiveFusion(nn.Module):
    def __init__(self, degradation_dim=3, features=64):
        super().__init__()

        self.weight_generator = nn.Sequential(
            nn.Linear(degradation_dim, 32),
            nn.ReLU(inplace=True),
            nn.Linear(32, 3)
        )

        self.fusion_conv = nn.Sequential(
            nn.Conv2d(features * 3, features, 1),
            nn.ReLU(inplace=True),
            nn.Conv2d(features, features, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(
        self,
        cnn_features,
        vit_features,
        physics_features,
        degradation
    ):
        weights = F.softmax(
            self.weight_generator(degradation),
            dim=1
        )

        w_cnn = weights[:, 0].view(-1, 1, 1, 1)
        w_vit = weights[:, 1].view(-1, 1, 1, 1)
        w_physics = weights[:, 2].view(-1, 1, 1, 1)

        cnn_weighted = cnn_features * w_cnn
        vit_weighted = vit_features * w_vit
        physics_weighted = physics_features * w_physics

        combined = torch.cat(
            [cnn_weighted, vit_weighted, physics_weighted],
            dim=1
        )

        fused = self.fusion_conv(combined)

        return fused, weights


# =========================================================
# 5. COMPLETE AQUALUME MODEL
# =========================================================
class AquaLume(nn.Module):
    def __init__(self):
        super().__init__()

        self.cnn = AquaLumeCNN()
        self.vit = AquaLumeViT()
        self.physics = AquaLumePhysics()
        self.adaptive_fusion = AdaptiveFusion()

        self.decoder = nn.Sequential(
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Upsample(
                scale_factor=2,
                mode="bilinear",
                align_corners=False
            ),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Upsample(
                scale_factor=2,
                mode="bilinear",
                align_corners=False
            ),
            nn.Conv2d(64, 32, 3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 3, 3, padding=1),
            nn.Sigmoid()
        )

    def forward(self, image, degradation):

        cnn_features = self.cnn(image)

        cnn_features = F.interpolate(
            cnn_features,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        vit_features = self.vit(image)

        physics_features = self.physics(
            image, degradation
        )

        fused_features, weights = self.adaptive_fusion(
            cnn_features,
            vit_features,
            physics_features,
            degradation
        )

        enhanced_image = self.decoder(fused_features)

        return enhanced_image, weights


print("All AquaLume v1 architecture classes defined successfully.")
```

```python
import numpy as np
import cv2
import torch
from PIL import Image

IMAGE_SIZE = 256

def compute_degradation_features(image):
    """
    Extract the 3 degradation features used by AquaLume v1.
    image: RGB image, values in [0, 1]
    """
    image = np.asarray(image, dtype=np.float32)
    image = np.clip(image, 0, 1)

    # 1. Color imbalance
    channel_means = image.mean(axis=(0, 1))
    color_imbalance = np.std(channel_means)

    # 2. Contrast
    image_uint8 = (image * 255).astype(np.uint8)
    gray = cv2.cvtColor(image_uint8, cv2.COLOR_RGB2GRAY)
    contrast = np.std(gray)

    # 3. Sharpness / Laplacian variance
    laplacian_variance = cv2.Laplacian(
        gray, cv2.CV_64F
    ).var()

    sharpness = np.log1p(laplacian_variance)

    raw_features = np.array(
        [color_imbalance, contrast, sharpness],
        dtype=np.float32
    ).reshape(1, -1)

    scaled_features = feature_scaler.transform(raw_features)

    return scaled_features.squeeze(0).astype(np.float32)


def enhance_underwater(image, return_weights=False):
    """
    AquaLume v1 inference.
    Accepts a PIL image and returns an enhanced PIL image.
    """

    image = image.convert("RGB")

    # Preserve original dimensions for final output
    original_size = image.size

    # Resize input to model resolution
    resized_image = image.resize(
        (IMAGE_SIZE, IMAGE_SIZE),
        Image.Resampling.BILINEAR
    )

    image_array = np.asarray(
        resized_image, dtype=np.float32
    ) / 255.0

    # Extract degradation descriptor
    degradation = compute_degradation_features(image_array)

    # Tensor preparation
    image_tensor = torch.from_numpy(
        image_array.transpose(2, 0, 1)
    ).unsqueeze(0).float().to(device)

    degradation_tensor = torch.from_numpy(
        degradation
    ).unsqueeze(0).float().to(device)

    # Inference
    with torch.no_grad():
        enhanced_tensor, weights = model(
            image_tensor, degradation_tensor
        )

    enhanced_array = (
        enhanced_tensor.squeeze(0)
        .cpu()
        .numpy()
        .transpose(1, 2, 0)
    )

    enhanced_array = np.clip(enhanced_array, 0, 1)

    enhanced_image = Image.fromarray(
        (enhanced_array * 255).astype(np.uint8)
    )

    # Resize enhanced output back to original dimensions
    enhanced_image = enhanced_image.resize(
        original_size, Image.Resampling.BILINEAR
    )

    weights_array = weights.squeeze(0).cpu().numpy()

    if return_weights:
        return enhanced_image, weights_array

    return enhanced_image


print("AquaLume inference pipeline created successfully.")
```

```python
import torch
import torch.nn.functional as F
import numpy as np
from PIL import Image

def enhance_with_branch_ablation(image, disabled_branch=None):

    image = image.convert("RGB")
    original_size = image.size

    resized = image.resize(
        (IMAGE_SIZE, IMAGE_SIZE),
        Image.Resampling.BILINEAR
    )

    image_array = np.asarray(resized, dtype=np.float32) / 255.0

    degradation = compute_degradation_features(image_array)

    image_tensor = torch.from_numpy(
        image_array.transpose(2, 0, 1)
    ).unsqueeze(0).float().to(device)

    degradation_tensor = torch.from_numpy(
        degradation
    ).unsqueeze(0).float().to(device)

    with torch.no_grad():

        cnn_features = model.cnn(image_tensor)

        cnn_features = F.interpolate(
            cnn_features,
            size=(64, 64),
            mode="bilinear",
            align_corners=False
        )

        vit_features = model.vit(image_tensor)

        physics_features = model.physics(
            image_tensor, degradation_tensor
        )

        # Zero the selected branch at inference
        if disabled_branch == "CNN":
            cnn_features = torch.zeros_like(cnn_features)

        elif disabled_branch == "ViT":
            vit_features = torch.zeros_like(vit_features)

        elif disabled_branch == "Physics":
            physics_features = torch.zeros_like(physics_features)

        fused_features, weights = model.adaptive_fusion(
            cnn_features,
            vit_features,
            physics_features,
            degradation_tensor
        )

        enhanced_tensor = model.decoder(fused_features)

    enhanced_array = (
        enhanced_tensor.squeeze(0)
        .cpu()
        .numpy()
        .transpose(1, 2, 0)
    )

    enhanced_array = np.clip(enhanced_array, 0, 1)

    enhanced_image = Image.fromarray(
        (enhanced_array * 255).astype(np.uint8)
    )

    enhanced_image = enhanced_image.resize(
        original_size,
        Image.Resampling.BILINEAR
    )

    return enhanced_image


print("Inference-time ablation function created.")
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image

# Original image
original = sample_image.convert("RGB")

# AquaLume
aqualume = full_output.convert("RGB")

# CLAHE (applied to luminance in LAB color space)
img_np = np.array(original)
lab = cv2.cvtColor(img_np, cv2.COLOR_RGB2LAB)
l, a, b = cv2.split(lab)

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
l_enhanced = clahe.apply(l)

clahe_img = cv2.cvtColor(
    cv2.merge((l_enhanced, a, b)),
    cv2.COLOR_LAB2RGB
)
clahe_img = Image.fromarray(clahe_img)

# GrayWorld white balance
img_float = img_np.astype(np.float32)
channel_means = img_float.mean(axis=(0, 1))

gray_mean = channel_means.mean()
balanced = img_float * (gray_mean / (channel_means + 1e-8))
balanced = np.clip(balanced, 0, 255).astype(np.uint8)

grayworld_img = Image.fromarray(balanced)

# Final comparison gallery
images = [
    ("Original", original),
    ("AquaLume", aqualume),
    ("CLAHE", clahe_img),
    ("GrayWorld", grayworld_img)
]

plt.figure(figsize=(16, 9))

for i, (title, img) in enumerate(images):
    plt.subplot(2, 2, i + 1)
    plt.imshow(img)
    plt.title(title, fontsize=14)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
